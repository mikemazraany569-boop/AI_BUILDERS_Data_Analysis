Why are the numbers the same each run?

Because of:np.random.seed(7)

seed() sets the starting point of the random number generator.

Normally, random numbers are generated using an internal changing state, so every run gives different results.

When you set:np.random.seed(7)

you tell NumPy:

"Start generating random numbers from the same starting position every time."

Therefore, the sequence of "random" numbers is reproducible.
