```python
mat1
```

```python
mat2
```

```python
np.dot(mat1,mat2)
```

```python
mat3=np.random.randint(10,size=(4,3))
mat4=np.random.randint(10,size=(4,3))
mat3.shape,mat4.shape
```

```python
np.dot(mat3.T,mat4)
```
