```python
import pandas as pd
cars_sales=pd.read_csv(r"C:\Users\HP\Downloads\cars_sales.csv")
cars_sales.head()
```

```python
cars_sales.dtypes
```

```python
cars_sales["sales date"]=pd.date_range("1/1/2026",periods=len(cars_sales))
cars_sales
```

```python
cars_sales["total sales"]=cars_sales["price"].astype(int).cumsum()
cars_sales
```

```python
cars_sales.plot(x="sales date",y="total sales")
```

```python
cars_sales["price"]=cars_sales["price"].astype(int)
cars_sales.plot(x="odometer(km)",y="price",kind="scatter")
```

```python
cars_sales.plot(x="name",y="odometer(km)",kind="bar")
```

```python
cars_sales["odometer(km)"].plot(kind="hist")
```

```python
cars_sales["odometer(km)"].plot.hist(bins=20)
```
