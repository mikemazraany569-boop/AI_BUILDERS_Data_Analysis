```python
import numpy as np
x=np.linspace(0,10,100)
x
```

```python
import matplotlib.pyplot as plt
```

```python
fig,ax=plt.subplots()
ax.plot(x,x**2)
```

```python
fig,ax=plt.subplots()
ax.scatter(x,np.exp(x))
```

```python
fig,ax=plt.subplots()
ax.scatter(x,np.sin(x))
```

```python
nb_of_students={"Uni Students":10,
                "High School Students":20,
                "Fresh Graduates":15}
fig,ax=plt.subplots()
ax.bar(nb_of_students.keys(),nb_of_students.values())                
```

```python
fig,ax=plt.subplots()
ax.barh(nb_of_students.keys(),nb_of_students.values())                
```

```python
x=np.random.randn(1000)
fig,ax=plt.subplots()
ax.hist(x,color="red")
```

```python
plt.plot([1,2,3,4])
```

```python
x=[1,2,3,4]
y=[11,12,13,14]
plt.plot(x,y)
```

```python
plt.plot(x,y,marker='+',color="orange")
plt.title("simple plot")
```

```python
fig,((ax1,ax2),(ax3,ax4))=plt.subplots(nrows=2,ncols=2,figsize=(10,5))
ax1.plot(x,np.exp(x))
ax2.plot(np.random.random(10),np.random.random(10))
ax3.barh(nb_of_students.keys(),nb_of_students.values())  
ax4.hist(np.random.randn(1000))
```
